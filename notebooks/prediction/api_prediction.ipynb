```python
import requests
```

```python
# The URL where your FastAPI app is running
# Locally run it with "uvicorn src.api.app:app --reload"
# Note: /forecast is the endpoint defined in the app.py
URL = "http://127.0.0.1:8000/forecast"

# This matches your ForecastRequest(BaseModel) schema
payload = {
    "weeks": 24
}
```

```python
try:
    # Send a POST request with the JSON payload
    response = requests.post(URL, json=payload)
    
    # Check if the request was successful
    if response.status_code == 200:
        data = response.json()
        print(f"✅ Prediction received for {data['horizon']} weeks:")
        print(data['forecast'])
    else:
        print(f"❌ Error {response.status_code}: {response.text}")

except requests.exceptions.ConnectionError:
    print("Failed to connect to the server. Make sure the FastAPI app is running.")
```
